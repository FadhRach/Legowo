```python
bawang_merah = "bawang_merah_random_submission.csv"
bawang_putih = "bawang_putih_bonggol_random_submission.csv"
beras_medium = "beras_medium_random_submission.csv"
beras_premium = "beras_premium_random_submission.csv"
cabai_merah_keriting = "cabai_merah_keriting_submission.csv"
cabai_rawit_merah = "cabai_rawit_merah_submission.csv"
daging_ayam_ras = "daging_ayam_ras_submission.csv"
daging_sapi_murni = "daging_sapi_murni_submission.csv"
gula_konsumsi = "gula_konsumsi_submission.csv"
minyak_goreng_curah = "minyak_goreng_curah_random_submission_updated.csv"
minyak_goreng_kemasan_sederhana = "minyak_goreng_kemasan_sederhana_random_submission_updated.csv"
telur_ayam_ras = "telur_ayam_ras_random_submission_updated.csv"
tepung_terigu_curah = "tepung_terigu_curah_random_submission_updated.csv"

file_paths = [
    bawang_merah, bawang_putih, beras_medium, beras_premium,
    cabai_merah_keriting, cabai_rawit_merah, daging_ayam_ras,
    daging_sapi_murni, gula_konsumsi, minyak_goreng_curah,
    minyak_goreng_kemasan_sederhana, telur_ayam_ras, tepung_terigu_curah
]

df_list = [pd.read_csv(file) for file in file_paths]
df_combined = pd.concat(df_list)

df_combined = df_combined.sort_values(by="id").reset_index(drop=True)

df_combined.to_csv("combined_random_dataset.csv", index=False)
```
